```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
coil20 = pd.read_csv('../coil20_2024.04.24-16.21.15.csv')
for n_classes in coil20['n_classes'].unique():
    coil20_part = coil20[coil20['n_classes'] == n_classes]
    acc_avg = coil20_part['acc'].mean()
    nmi_avg = coil20_part['nmi'].mean()
    print(f'{n_classes} classes, acc = {acc_avg:.3f}, nmi = {nmi_avg:.3f}')
    print()
```
