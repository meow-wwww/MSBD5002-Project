```python
from SymANLS_pivot import SymANLS_pivot
import cifar10
import reuters
import coil20
import tdt2

import numpy as np
import matplotlib.pyplot as plt
```

```python
import random
random.seed(42)
```

# Coil20

## CLUSTER_NUM = 5

```python
CLUSTER_NUM = 5
r = 5 # equal to the number of clusters
maxiter = 500
lambda_values = [0.01, 0.1, 0.5, 1, 5, 10, 39.9]
```

```python
mat, truelabel = coil20.read_data(num_of_class=CLUSTER_NUM, random_shuffle=True, 
                              random_choose_class=True)
print("*" * 30)
print(f'[Using Classes] {np.unique(np.array(truelabel))}')
print("*" * 30)

# cosine similarity matrix
mat -= np.mean(mat)
norm = np.linalg.norm(mat, axis=1).reshape(-1, 1)
mat_normalized = mat / norm
X = mat_normalized @ mat_normalized.T

U0 = np.random.rand(X.shape[0], r)
V0 = U0.copy()

ERRORS = []
for lambda_ in lambda_values:
    print(f'[Current Lambda] {lambda_}')
    U, V, e, t, diff, NMI, ACC = SymANLS_pivot(X, U0, V0, maxiter, lambda_, truelabel)
    ERRORS.append(e)
```

```python
import math
iterations = range(0, maxiter + 2)  
log_errors = [[math.log(e) for e in error_list] for error_list in ERRORS]

plt.figure()
for i, log_e in enumerate(log_errors):
    plt.plot(iterations, log_e, label=f"lambda = {lambda_values[i]}")

plt.xlabel('Iteration k')
plt.ylabel('Log Error E_k')
plt.legend()
plt.show()
```
