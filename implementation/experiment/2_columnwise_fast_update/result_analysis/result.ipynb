```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
coil20 = pd.read_csv('../coil20.csv')
for n_classes in coil20['n_classes'].unique():
    coil20_part = coil20[coil20['n_classes'] == n_classes]
    for lambda_ in coil20_part['lambda'].unique():
        acc_avg = coil20_part[coil20_part['lambda'] == lambda_]['acc'].mean()
        nmi_avg = coil20_part[coil20_part['lambda'] == lambda_]['nmi'].mean()
        print(f'{n_classes} classes, lambda = {lambda_}, acc = {acc_avg:.3f}, nmi = {nmi_avg:.3f}')
    print()
```
