```python
import numpy as np
```

```python
import coil20
mat, label = coil20.read_data() 
# by default, read all classes, no random shuffle
mat.shape, label.shape
```

```python
import cifar10
mat, label = cifar10.read_data(num_of_class=3) 
# only read 3 classes (the first 3 classes), no random shuffle
mat.shape, label.shape
```

```python
import reuters
mat, label = reuters.read_data(num_of_class=3, random_choose_class=True)
# only read 3 classes (randomly choose 3 classes), no random shuffle
# in text data, different classes may have different number of samples, 
# so the number of samples returned may be different across different runs
print(mat.shape, label.shape)
print(np.unique(label, return_counts=True))
```

```python
import tdt2
mat, label = tdt2.read_data(num_of_class=3, random_choose_class=True, random_shuffle=True)
# only read 3 classes (randomly choose 3 classes), random shuffle
# in text data, different classes may have different number of samples,
# so the number of samples returned may be different across different runs
print(mat.shape, label.shape)
print(np.unique(label, return_counts=True))
print(label[:20])
```
